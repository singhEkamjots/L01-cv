````python
### Q1

## 1.1: Run the code
import pickle
with open('minst.pkl', 'rb') as f: # adjusted the code to open the specific file in the zip
    data = pickle.load(f)
X_train, y_train = data['X_train'], data['y_train']
X_test, y_test = data['X_test'], data['y_test']

import matplotlib.pyplot as plt
import numpy as np
np.set_printoptions(edgeitems=30, linewidth=100000)
for i in range(5):
    print(y_test[i],'\n') # Print the label
    print(X_test[i],'\n') # Print the matrix of values
    plt.contourf(np.rot90(X_test[i].transpose())) # Make a contour plot of the matrix values
    plt.show()
````

````python
### Q2

## 2.1: What is the shape of X_train and X_test?
print(X_train.shape)
print(X_test.shape)
# (60000, 28, 28) X_train has 60,000 images and they are each 28x28 pixels.
# (10000, 28, 28) X_test has 10,000 images are they are each 28x28 pixels.

## 2.2: What is the shape of X_train[i] and X_test[i] for each index i?
print(X_train[0].shape)
print(X_test[0].shape)
# (28, 28)
# (28, 28)
# Both have the same dimensions of 28x28.

## 2.3: What is the shape of y_train and y_test?
print(y_train.shape)
print(y_test.shape)
# (60000,) y_train has 60,000 images and is one-dimensional.
# (10000,) y_test has 10,000 images and is one-dimensional.


### Q3

## 3.1 and 3.2: Use Numpy's .reshape() method to covert the training and testing data from a matrix into an vector of features.
X_train_reshaped = X_train.reshape((60000, 784))
X_test_reshaped = X_test.reshape((10000, 784))

## 3.3: So, X_test[index].reshape((1,784)) will convert the  𝑖𝑛𝑑𝑒𝑥 -th element of X_test into a  28×28=784 -length row vector of values, rather than a matrix.
## Turn X_train into an  𝑁×784  matrix  𝑋  that is suitable for scikit-learn's kNN classifier where  𝑁  is the number of observations and  784=28∗28
#(you could use, for example, a for loop).
X = []

for i in range(len(X_train)):
    X.append(X_train[i].reshape(784))

X = np.array(X)

print(X.shape)


### Q4

## 4.1: Use the reshaped X_test and y_test data to create a  𝑘 -nearest neighbor classifier of digit.
from sklearn.neighbors import KNeighborsClassifier

model = KNeighborsClassifier(n_neighbors=3)
model.fit(X, y_train)

## 4.2: What is the optimal number of neighbors  𝑘 ? If you can't determine this, play around with different values of  𝑘  for your classifier.
# for k in range(1, 8): # *took a long time to run
    #model = KNeighborsClassifier(n_neighbors=k)
    #model.fit(X, y_train)
    #print(k, model.score(X_test_reshaped, y_test))
# k=3 is the optimal number of neighbor with an accuracy of 97.05% on the test set.


### Q5

## 5.1: For the optimal number of neighbors, how well does your predictor perform on the test set? Report the accuracy.
# Using k=3, it got an accuracy of 97.05% on the test set.

## 5.2: Compute a confusion matrix.
from sklearn.metrics import ConfusionMatrixDisplay

y_pred = model.predict(X_test_reshaped)
ConfusionMatrixDisplay.from_predictions(y_test, y_pred, normalize='true') # normalized it to see proportions
plt.show()

## 5.3: Explain your findings.
# The kNN classifier performed well with a 97.05% accuracy. Based on the confusion matrix, you can see that most values are concentrated along the diagonal.
# The model performed the best on the digits 0, 1, and 6 with a 99%, 100%, and 99% accuracy, respectively.

### Q6

## 6.1: For your confusion matrix, which mistakes are most likely?
#The most common mistakes involve the model confusing 7s with 1s (2% of the actual 7s), 4s with 9s (1.9% of the 4s), and 3s with 5s (1.6% of the threes). These mistakes make sense because the handwritten versions of these numbers can look very similar depending on how someone writes them.

## 6.2: Do you find any interesting patterns?
# 1 has a basically 100% recall, making it the easiest to recognize, while digit 8 has the lowest recall at 94%, making it the hardest to recognize. What's interesting is that misclassifications are not always symmetrical. Misclassifications are not always symmetrical. For example, 7s are sometimes classified as 1s, but 1s are almost never classified as 7s.

### Q7

## 7.1: So, this is how computers "see." They convert an image into a matrix of values, that matrix becomes a vector in a dataset,
## and then we deploy ML tools on it as if it was any other kind of tabular data. To make sure you follow this,
## invent a way to represent a color photo in matrix form.
# A color photograph could be represented as a matrix where each pixel contains three numerical values representing red, green, and blue (RGB) intensity on a scale of 0–100. A value of 0 would represent no intensity of that color, while 100 would represent the maximum intensity. For example, a pixel with RGB values of [100, 0, 0] would be red, while [0, 0, 100] would be blue. A 28×28 photograph could therefore be represented as a 28×28×3 array.


## 7.2: Describe how you could convert it into tabular data. (Hint: RGB color codes provide a method of encoding a numeric value that represents a color.)
# Converting photograph into the tabular data would mean I would turn the matrix into a single vector containing the three RGB values for every pixel. For a 28×28 image, this would produce 2,352 features (28×28×3). Each photograph would represent one row in a dataset, with each column containing the intensity of one color channel for a particular pixel.

````
